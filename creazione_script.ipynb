## Dataframe churned user

```python
#valore base
first = np.random.randint(10,201)

#valori attorno al valore base per i primi 6 mesi
rumore = np.random.normal(0, first * .1, 6)
first_6_values = np.round(first + rumore).astype(int)

#valori calanti ultimi 6 mesi
limit = max(round(first * .1), 6)
valore = first
risultati = []
for i in range(6):
    sottrazione = np.random.randint(5,limit + 1)
    valore = max(valore - sottrazione, 0)   # mai sotto zero
    risultati.append(valore)

#unisco array + lista 
logins = np.append(first_6_values, risultati)


churned_signal = pd.DataFrame({
    "customerID": [np.random.randint(100000, 1000000)] * 12
    , "month": np.arange(-12,0)
    , "logins": logins
    , "churned": True
})

churned_signal
```

## Churned users with signal Dataframe

```python
#dataframe list
df_list = []

for a in range(15):
    #valore base
    first = np.random.randint(10,201)

    # Valori attorno al valore base per i primi 6 mesi
    rumore = np.random.normal(0, first * .1, 6) #ne genera 6
    first_6_values = np.round(first + rumore).astype(int) #aggiunge rumore e rende int.

    #valori calanti ultimi 6 mesi
    limit = max(round(first * .1), 6)
    valore = first
    risultati = [] #lista ultimi 6 mesi
    for i in range(6):
        sottrazione = np.random.randint(5,limit + 1)
        valore = max(valore - sottrazione, 0)   # mai sotto zero
        risultati.append(valore)


# Correzione: ogni mese il cliente perde una PERCENTUALE casuale del valore
#attuale (es. tra 5% e 15%), invece di un numero fisso.


    #unisco array + lista 
    logins = np.append(first_6_values, risultati)


    churned_user = pd.DataFrame({
        "customerID": [np.random.randint(100000, 1000000)] * 12
        , "month": np.arange(-12,0)
        , "logins": logins
        , "churned": True
    })
    #store new df in list
    df_list.append(churned_user)

churned_signal = pd.concat(df_list)
churned_signal.reset_index(drop=True, inplace=True)
```
